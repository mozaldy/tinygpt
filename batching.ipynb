```python

X = [1, 2, 3, 4]
Y = [2, 4, 6, 8]

w = 0.0
b = 0.0
learning_rate = 0.05
batch_size = 2
epochs = 10


for j in range(epochs):
    
    for i in range(0, len(X), batch_size):
        mse = 0
        xs = X[i:i+batch_size]
        ys = Y[i:i+batch_size]

        
        dw = 0
        db = 0
        for x, y in zip(xs, ys):
            predict = w*x+b
            dw += (predict - y) * x
            db += (predict - y)
        
        dw *= 2/batch_size
        db *= 2/batch_size

        w = w - learning_rate * dw
        b = b - learning_rate * db

    full_mse = 0
    for x, y in zip(X, Y):
        predict = w * x + b
        full_mse += (predict - y) ** 2

    full_mse /= len(X)

    
        
    print("="*20)
    print(f"Epoch {j + 1}: w = {w:.2f}, b = {b:.2f}")
    print(f"dw: {dw:.2f}, db: {db:.2f}")
    print(f"Full-dataset MSE: {full_mse:.2f}")
    
    
```
