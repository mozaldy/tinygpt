# Stage 5 — Mini-batch Training

Batch size 2, learning rate 0.05, ten epochs. After each epoch, compute the full-dataset MSE using final parameters. Run cells in order. See [the note](../notes/06-mini-batch-gradient-descent.md).

```python

X = [1, 2, 3, 4]
Y = [2, 4, 6, 8]

w = 0.0
b = 0.0
learning_rate = 0.05
batch_size = 2
epochs = 10
loss_history = []


for j in range(epochs):
    
    for i in range(0, len(X), batch_size):
        xs = X[i:i+batch_size]
        ys = Y[i:i+batch_size]

        
        dw = 0
        db = 0
        for x, y in zip(xs, ys):
            predict = w*x+b
            dw += (predict - y) * x
            db += (predict - y)
        
        dw *= 2/len(xs)
        db *= 2/len(xs)

        w = w - learning_rate * dw
        b = b - learning_rate * db

    full_mse = 0
    for x, y in zip(X, Y):
        predict = w * x + b
        full_mse += (predict - y) ** 2

    full_mse /= len(X)
    loss_history.append(full_mse)

    
        
    print("="*20)
    print(f"Epoch {j + 1}: w = {w:.2f}, b = {b:.2f}")
    print(f"dw: {dw:.2f}, db: {db:.2f}")
    print(f"Full-dataset MSE: {full_mse:.2f}")
    
    

```

## Full-dataset MSE after each epoch

```python
import matplotlib.pyplot as plt
plt.plot(range(1, epochs + 1), loss_history, marker='o')
plt.xlabel('Epoch')
plt.ylabel('Full-dataset MSE')
plt.grid(True)
plt.show()
```
