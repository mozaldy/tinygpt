# 1. Forward Pass

```python

w = 3.0
x = 2.0
b = 1.0

# Step 1: Calculate the model output
a = w * x + b

# Step 2: Calculate the loss
L = a ** 2

print("Output:", a)
print("Loss:", L)

```

# 2. Backpropagation

```python
dL_dw = 2 * a * x
dL_db = 2 * a

print("Weight gradient:", dL_dw)
print("Bias gradient:", dL_db)
```

# 3. Gradient Descent

```python
# Single iteration version
lr = 0.01
w = w - lr * dL_dw
b = b - lr * dL_db

# Step 1: Calculate the model output
a = w * x + b

# Step 2: Calculate the loss
L = a ** 2

print("After 1 iteration")
print("Output:", a)
print("Loss:", L)
```

# 4. Looped iteration learning

```python



w = 3.0
x = 2.0
b = 1.0
lr = 0.1

loss_history = []

for i in range(10):
    print(f"Iteration {i}")
    # 1. Forward pass
    a = w * x + b
    L = a ** 2

    loss_history.append(L)

    print("Weight:", w)
    print("Bias:", b)
    print("Loss:", L)
    # 2. Backpropagation
    dL_dw = 2 * a * x
    dL_db = 2 * a

    # 3. Gradient descent
    w = w - lr * dL_dw
    b = b - lr * dL_db

  
```

```python
import matplotlib.pyplot as plt

plt.plot(loss_history)
plt.xlabel("Iteration")
plt.ylabel("Loss")
plt.title("Training Loss")
plt.show()
```
